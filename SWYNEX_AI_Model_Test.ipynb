```python

import pandas as pd

data = pd.read_csv("customer_feedback.csv")

print("Dataset loaded successfully!")
print("Total feedback:", len(data))

data.head()
```

```python

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

# Create AI model
model = Pipeline([
    ("tfidf", TfidfVectorizer(ngram_range=(1, 2))),
    ("classifier", LogisticRegression(max_iter=1000))
])

# Train AI model
model.fit(data["feedback"], data["sentiment"])

print("AI Model trained successfully!")
print("Training samples:", len(data))
print("Sentiment categories:", list(model.classes_))
```

```python

# Test AI model with new customer feedback

test_feedback = [
    "The product is excellent",
    "Very bad customer service",
    "The product is okay",
    "I am very happy with the service",
    "The delivery was terrible"
]

print("AI CUSTOMER FEEDBACK CLASSIFICATION")
print("-----------------------------------")

for feedback in test_feedback:
    prediction = model.predict([feedback])[0]

    print("Customer Feedback:", feedback)
    print("AI Prediction:", prediction)
    print("-----------------------------------")
```

```python

# Check the sentiment distribution in our dataset

print("DATASET ANALYSIS")
print("-----------------------")

print(data["sentiment"].value_counts())

print("\nTotal samples:", len(data))
print("Missing values:")
print(data.isnull().sum())
```

```python

from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

# Improved AI Model
improved_model = Pipeline([
    ("tfidf", TfidfVectorizer(
        analyzer="char_wb",
        ngram_range=(2, 5)
    )),
    ("classifier", LogisticRegression(
        max_iter=2000,
        C=10
    ))
])

# Train model
improved_model.fit(data["feedback"], data["sentiment"])

# Test predictions
test_feedback = [
    "The product is excellent",
    "Very bad customer service",
    "The product is okay",
    "I am very happy with the service",
    "The delivery was terrible"
]

print("IMPROVED AI MODEL RESULTS")
print("------------------------------")

for feedback in test_feedback:
    prediction = improved_model.predict([feedback])[0]
    print("Feedback:", feedback)
    print("Prediction:", prediction)
    print("------------------------------")
```
